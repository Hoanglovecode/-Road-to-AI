```python
df=pd.read_csv("Titanic-Dataset 2.csv")
```

```python
cols = ['Name', 'Pclass', 'Age', 'Fare']

# Sắp xếp tuổi tăng dần (mặc định ascending=True):
df_age_asc = df.sort_values(by='Age')[cols]
print("Top 5 người trẻ nhất:")
print(df_age_asc.head().to_string(index=False))

# Sắp xếp giá vé giảm dần (người chi nhiều tiền nhất lên đầu):
df_fare_desc = df.sort_values(by='Fare', ascending=False)[cols]
print("Top 5 người mua vé đắt nhất:")
print(df_fare_desc.head().to_string(index=False))
```

```python
# Hạng vé tăng dần (1, 2, 3), Tuổi trong từng hạng vé giảm dần:
df_multi_sort = df.sort_values(
    by=['Pclass', 'Age'],
    ascending=[True, False]
)[cols]

print("Sắp xếp Pclass tăng dần, Age giảm dần:")
print(df_multi_sort.head(10).to_string(index=False))
```

```python
# Ví dụ 1: Phân loại nhóm tuổi
def phan_loai_tuoi(tuoi):
    if pd.isna(tuoi):
        return "Không rõ"
    elif tuoi < 18:
        return "Trẻ em"
    elif tuoi < 60:
        return "Người lớn"
    else:
        return "Người cao tuổi"

df['Nhom_Tuoi'] = df['Age'].apply(phan_loai_tuoi)
print(df[['Age', 'Nhom_Tuoi']].head())

# Minh họa kết quả gán nhãn nhóm tuổi bằng apply (đặt hình ngoài khối mã)

# Ví dụ 2: Trích xuất danh xưng (Title) từ cột Họ tên
def extract_title(name):
    # Tách chuỗi theo dấu phẩy, sau đó lấy trước dấu chấm
    after_comma = name.split(',')[1]
    title = after_comma.split('.')[0].strip()
    return title

df['Title'] = df['Name'].apply(extract_title)
print(f"Top các danh xưng phổ biến:{df['Title'].value_counts().head(5)}")
```

Chỉ giữ các dòng có Fare không bị thiếu, sắp xếp theo PassengerId tăng dần. Tính rolling mean của Fare với cửa sổ 5 hành khách liên tiếp. Chỉ tính khi có đủ 5 giá trị.

```python
# ==========================================
# TODO: Xử lý Rolling Mean cho cột Fare
# ==========================================

# Bước 1: Lọc bỏ các dòng có giá trị Fare bị thiếu (NaN)
# Gợi ý: Dùng phương thức .dropna(subset=['Fare'])
df_clean = df.dropna(subset=['Fare']).copy()

# Bước 2: Sắp xếp DataFrame theo 'PassengerId' tăng dần
# Gợi ý: Dùng .sort_values(by='PassengerId', ascending=True)
df_clean = df_clean.sort_values(by='PassengerId', ascending=True)

# Bước 3: Tính Rolling Mean của 'Fare' với cửa sổ (window) = 5
# Yêu cầu: "Chỉ tính khi có đủ 5 giá trị" -> đặt min_periods=5
# Gợi ý: df_clean['Fare'].rolling(window=5, min_periods=5).mean()
df_clean['Fare_Rolling_Mean_5'] = df_clean['Fare'].rolling(window=5, min_periods=5).mean()

# Bước 4: Hiển thị kết quả kiểm tra
# (4 dòng đầu tiên sẽ nhận giá trị NaN do chưa đủ 5 dòng, từ dòng thứ 5 mới có giá trị mean)
df_clean[['PassengerId', 'Fare', 'Fare_Rolling_Mean_5']].head(10)
```

Loại các dòng thiếu Age hoặc Fare, sắp xếp theo Age tăng dần. Tính rolling mean của Fare với cửa sổ 10 dòng liên tiếp, chỉ tính khi đủ 10 dòng. Tìm dòng có rolling mean cao nhất.

Loại các dòng thiếu Age hoặc Fare, sắp xếp theo Pclass, Age, rồi PassengerId. Với từng Pclass riêng, tính rolling mean của Fare trên 5 dòng liên tiếp, chỉ tính khi đủ 5 dòng. Sau đó lọc các dòng có rolling mean lớn hơn 30 và so sánh giữa các hạng vé.

```python
import numpy as np
import pandas as pd
'''
RandomForestRegressor: mô hình học mối quan hệ giữa các feature.



IterativeImputer: lặp đi lặp lại quá trình dự đoán giá trị thiếu.



max_iter=10: tối đa 10 vòng lặp cập nhật.
'''
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.ensemble import RandomForestRegressor

df = pd.DataFrame({
    'Age': [18, 19, 20, 21, 22],
    'Height': [165, 170, np.nan, 175, 180],
    'Weight': [55, 60, 65, 70, 75]
})

imputer = IterativeImputer(
    estimator=RandomForestRegressor(random_state=42),
    max_iter=10,
    random_state=42
)

df_imputed = pd.DataFrame(
    imputer.fit_transform(df),
    columns=df.columns
)

print(df_imputed)
```
