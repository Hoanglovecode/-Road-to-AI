```python
from collections import deque

# 1. Định nghĩa Node cho Binary Tree
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def build_tree(values: list) -> TreeNode | None:
    """
    Xây dựng cây nhị phân từ mảng level-order (dùng None cho node rỗng).
    Ví dụ: [3, 9, 20, None, None, 15, 7]
    """
    if not values or values[0] is None:
        return None
    
    root = TreeNode(values[0])
    queue = deque([root])
    i = 1
    n = len(values)
    
    while queue and i < n:
        curr = queue.popleft()
        
        # Node con bên trái
        if i < n:
            if values[i] is not None:
                curr.left = TreeNode(values[i])
                queue.append(curr.left)
            i += 1
            
        # Node con bên phải
        if i < n:
            if values[i] is not None:
                curr.right = TreeNode(values[i])
                queue.append(curr.right)
            i += 1
            
    return root
```
