```python
def to_linked_list(elements: list[int]) -> ListNode | None:
    """Chuyển đổi một Python list thành Linked List."""
    dummy = ListNode(0)
    curr = dummy
    for val in elements:
        curr.next = ListNode(val)
        curr = curr.next
    return dummy.next

def to_list(head: ListNode | None) -> list[int]:
    """Chuyển đổi Linked List về lại Python list để kiểm tra."""
    result = []
    curr = head
    while curr:
        result.append(curr.val)
        curr = curr.next
    return result
```
