```python
def containsDuplicate(nums: list[int]) -> bool:
    empty=set()
    for i in nums:
        if i not in empty:
            empty.add(i)
        else:
            return True
    return len(empty)!=len(nums)
```

```python
assert containsDuplicate([1,2,3,1]) == True, "Test 1 Failed"
assert containsDuplicate([1,2,3,4]) == False, "Test 2 Failed"
assert containsDuplicate([1,1,1,3,3,4,3,2,4,2]) == True, "Test 3 Failed"
print("✅ Xuất sắc! Đã vượt qua tất cả test cases!")
```
