```python
def two_sum(nums: list[int], target: int) -> list[int]:
    array= [(num,i) for i,num in enumerate(nums)]
    array.sort()
    l=0
    r=len(array)-1
    while l<r:
        total=array[l][0]+array[r][0]
        if total<target:
            l+=1
        elif total>target:
            r-=1
        else:
            return [array[l][1],array[r][1]]
    return []

```

```python
# Test cases
assert sorted(two_sum([2, 7, 11, 15], 9)) == [0, 1], "Test 1 Failed"
assert sorted(two_sum([3, 2, 4], 6)) == [1, 2], "Test 2 Failed"
assert sorted(two_sum([3, 3], 6)) == [0, 1], "Test 3 Failed"
print("✅ Xuất sắc! Đã vượt qua tất cả test cases!")
```
