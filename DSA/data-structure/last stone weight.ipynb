```python
import heapq
def lastStoneWeight(stones: list[int]) -> int:
    """
    Mô phỏng quá trình va đập các hòn đá bằng Max-Heap.
    Độ phức tạp mong muốn: O(n log n) thời gian, O(n) không gian bộ nhớ.
    """
    maxheap = [-s for s in stones]
    heapq.heapify(maxheap)
    while len(maxheap) > 1:
        y = -heapq.heappop(maxheap)
        x = -heapq.heappop(maxheap)
        if y != x:
            heapq.heappush(maxheap, -(y - x))
    return -maxheap[0] if maxheap else 0
```

```python
# Test 1: Trường hợp cơ bản
# [2,7,4,1,8,1] -> đập 8 & 7 còn 1 -> còn [2,4,1,1,1] -> đập 4 & 2 còn 2 -> ... -> còn 1
assert lastStoneWeight([2, 7, 4, 1, 8, 1]) == 1, "Test 1 Failed"

# Test 2: Chỉ có 1 hòn đá duy nhất
assert lastStoneWeight([1]) == 1, "Test 2 Failed"

# Test 3: Hai hòn đá bằng nhau bị phá hủy hoàn toàn -> Trả về 0 (Edge case quan trọng)
assert lastStoneWeight([2, 2]) == 0, "Test 3 Failed"

# Test 4: Ba hòn đá
# [3, 7, 2] -> đập 7 & 3 còn 4 -> [4, 2] -> đập 4 & 2 còn 2
assert lastStoneWeight([3, 7, 2]) == 2, "Test 4 Failed"

print("✅ Xuất sắc! Đã vượt qua tất cả test cases!")
```
