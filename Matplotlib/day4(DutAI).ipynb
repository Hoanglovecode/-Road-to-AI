```python
import pandas as pd
import matplotlib.pyplot as plt

# reading the database
data = pd.read_csv("tips.csv")

# Scatter plot with day against tip
sc = plt.scatter(
    data['day'],
    data['tip'],
    c=data['size'],       # màu thể hiện size
    s=data['total_bill'], # kích thước thể hiện total_bill
    cmap='viridis'        # thêm colormap cho đẹp (tuỳ chọn)
)

# Thêm title
plt.title("Scatter Plot")

# Thêm nhãn trục
plt.xlabel('Day')
plt.ylabel('Tip')

# Thêm colorbar + nhãn cho colorbar
cbar = plt.colorbar(sc)
cbar.set_label('Party size')

# Thêm chú giải cho kích thước điểm (total_bill)
from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], marker='o', color='w', label='Small bill',  markersize=5,  markerfacecolor='gray'),
    Line2D([0], [0], marker='o', color='w', label='Medium bill', markersize=10, markerfacecolor='gray'),
    Line2D([0], [0], marker='o', color='w', label='Large bill',  markersize=15, markerfacecolor='gray'),
]
plt.legend(handles=legend_elements, title='Total bill', loc='upper left')

plt.show()
```

```python
# Scatter plot with day against tip
plt.plot(data['tip'])
plt.plot(data['size'])

# Adding Title to the Plot
plt.title("Scatter Plot")

# Setting the X and Y labels
plt.xlabel('Day')
plt.ylabel('Tip')

plt.show()
```

```python
# Bar chart with day against tip
plt.bar(data['day'], data['tip'])

plt.title("Bar Chart")

# Setting the X and Y labels
plt.xlabel('Day')
plt.ylabel('Tip')

# Adding the legends
plt.show()
```

```python
# histogram of total_bills
plt.hist(data['total_bill'], bins=20)

plt.title("Histogram")

# Adding the legends
plt.show()
```
