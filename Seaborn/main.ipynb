```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

tips = pd.read_csv('tips.csv')

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    size="size",
    palette="Set1"
)

plt.title("Total Bill vs Tip")
plt.show()
```

```python
sns.lineplot(
    data=tips,
    x="size",
    y="tip",
    hue="sex",
    marker="o"
)

plt.title("Tip by Group Size")
plt.show()
```

```python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    errorbar=None
)

plt.title("Average Bill by Day")
plt.show()
```

```python
sns.histplot(
    data=tips,
    x="total_bill",
    bins=20,
    kde=True
)

plt.title("Distribution of Total Bill")
plt.show()
```

```python
sns.boxplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)

plt.title("Tip Distribution by Day and Sex")
plt.show()
```

```python
sns.violinplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex",
    split=True
)

plt.title("Tip Distribution by Day and Sex")
plt.show()
```

```python
corr = tips[["total_bill", "tip", "size"]].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(
    tips,
    vars=["total_bill", "tip", "size"],
    hue="sex",
    diag_kind="hist"
)

plt.show()
```

```python
sns.countplot(
    data=tips,
    x="day",
    hue="sex"
)

plt.title("Number of Customers by Day")
plt.show()
```

```python
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="scatter"
)

plt.show()
```

```python
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    ci=95
)

plt.title("Regression between Bill and Tip")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.scatterplot(data=tips, x="total_bill", y="tip")

plt.title("Total Bill vs Tip")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    ax=ax
)

ax.set_title("Total Bill vs Tip")
ax.set_xlabel("Total Bill")
ax.set_ylabel("Tip")

plt.show()
```
