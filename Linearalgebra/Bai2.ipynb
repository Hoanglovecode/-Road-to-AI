```python
import numpy as np
A_np=np.array([[4, -2],[1, 1]])
eigenvalues,eigenvectors=np.linalg.eig(A_np)
n=A_np.shape[0]
D=np.zeros((n, n))
for i in range(n):
    D[i,i]=eigenvalues[i]
P=eigenvectors
P_inv=np.linalg.inv(P)
A_reconstructed=P@D@P_inv
print("Eigenvalues:\n", eigenvalues)
print("Eigenvectors (P):\n", P)
print("Diagonal matrix (D):\n", D)
print("Inverse matrix (P_inv):\n", P_inv)
print("Reconstructed matrix (P @ D @ P_inv):\n", A_reconstructed)
print("Original matrix (A):\n", A_np)
print("Reconstruction correct:", np.allclose(A_np, A_reconstructed))
```

```python
import numpy as np
X = np.array([[1, 2], [3, 4], [5, 6]], dtype=float)
m, n = X.shape
eigenvalues, V = np.linalg.eig(X.T @ X)
idx = np.argsort(eigenvalues)[::-1]
S = np.sqrt(np.maximum(eigenvalues[idx], 0))
V = V[:, idx]
U = X @ V
for i in range(len(S)):
    if S[i] > 1e-10:
        U[:, i] /= S[i]
VT = V.T
print("U:\n", U)
print("S:\n", S)
print("VT:\n", VT)
```
