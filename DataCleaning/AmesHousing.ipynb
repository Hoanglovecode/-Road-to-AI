```python
import numpy as np
from sklearn.compose import ColumnTransformer, make_column_selector
from sklearn.pipeline import Pipeline

# 1. Numerical pipeline
numeric_pipeline = Pipeline(
    [("imputer", SimpleImputer(strategy="median", add_indicator=True))]
)

# 2. Categorical pipeline
categorical_pipeline = Pipeline(
    [("imputer", SimpleImputer(strategy="most_frequent", add_indicator=True))]
)

# 3. Automatically identify column types
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_pipeline, make_column_selector(dtype_include=np.number)),
        ("cat", categorical_pipeline, make_column_selector(dtype_exclude=np.number)),
    ]
)

# 4. Fit and transform
X_imputed = preprocessor.fit_transform(df)

# 5. Convert back to DataFrame
df_imputed = pd.DataFrame(
    X_imputed, columns=preprocessor.get_feature_names_out(), index=df.index
)

print(df_imputed.head())
```

```python
# 3. Phương pháp 2: Z-Score truyền thống
mean = numeric.mean()
std = numeric.std()
z_scores = (numeric - mean) / std
z_outliers = df[np.abs(z_scores) > 3]
```

```python
# 4. Phương pháp 3: Modified Z-Score (MAD)
median = numeric.median()
mad = np.median(np.abs(numeric - median))
modified_z_scores = 0.6745 * (numeric - median) / mad
mod_z_outliers = df[np.abs(modified_z_scores) > 3.5]
```

```python
# Chỉ lấy numerical features
# Loại những cột định danh và mã phân loại không có ý nghĩa định lượng
numeric = df.select_dtypes(include="number").drop(
    columns=["Id", "PID", "Order", "MS SubClass"],
    errors="ignore"
)

# ==============================
# 1. IQR METHOD
# ==============================

Q1 = numeric.quantile(0.25)
Q3 = numeric.quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

iqr_mask = numeric.lt(lower) | numeric.gt(upper)

# ==============================
# 2. Z-SCORE
# ==============================

mean = numeric.mean()
std = numeric.std().replace(0, np.nan)

z_scores = (numeric - mean) / std

z_mask = z_scores.abs() > 3

# ==============================
# 3. MODIFIED Z-SCORE (MAD)
# ==============================

median = numeric.median()

# MAD riêng cho từng feature
mad = numeric.sub(median).abs().median(axis=0)

modified_z_scores = (
    0.6745 * numeric.sub(median) / mad.replace(0, np.nan)
)

mod_z_mask = modified_z_scores.abs() > 3.5

# ==============================
# 4. CREATE REPORT
# ==============================

outlier_report = pd.DataFrame({
    "IQR_Count": iqr_mask.sum(),
    "IQR_Percent": iqr_mask.mean() * 100,

    "ZScore_Count": z_mask.sum(),
    "ZScore_Percent": z_mask.mean() * 100,

    "ModifiedZ_Count": mod_z_mask.sum(),
    "ModifiedZ_Percent": mod_z_mask.mean() * 100
})

print(outlier_report.round(2))

# Số dòng có ít nhất một feature bị đánh dấu
print("IQR flagged rows:", iqr_mask.any(axis=1).sum())
print("Z-score flagged rows:", z_mask.any(axis=1).sum())
print("Modified Z flagged rows:", mod_z_mask.any(axis=1).sum())
```

```python

from pathlib import Path

# Publication settings
sns.set_theme(style="whitegrid", context="notebook")

plt.rcParams.update({
    "figure.dpi": 120,
    "savefig.dpi": 300,
    "axes.titlesize": 15,
    "axes.labelsize": 12,
    "font.size": 11,
    "axes.spines.top": False,
    "axes.spines.right": False
})

Path("eda_figures").mkdir(exist_ok=True)

# Automatically find columns despite spaces/underscores
def find_col(data, *candidates):
    lookup = {
        "".join(ch.lower() for ch in str(col) if ch.isalnum()): col
        for col in data.columns
    }

    for candidate in candidates:
        key = "".join(
            ch.lower() for ch in candidate if ch.isalnum()
        )

        if key in lookup:
            return lookup[key]

    raise KeyError(
        f"Cannot find {candidates}. "
        f"Available columns: {data.columns.tolist()}"
    )


sale = find_col(df, "SalePrice")
neighborhood = find_col(df, "Neighborhood")
kitchen = find_col(df, "KitchenQual", "Kitchen Qual")
area = find_col(df, "GrLivArea", "Gr Liv Area")
quality = find_col(df, "OverallQual", "Overall Qual")
garage = find_col(df, "GarageCars", "Garage Cars")

print("Selected columns:")
print(sale, neighborhood, kitchen, area, quality, garage)
```

Biểu đồ 1 – SalePrice Distribution

```python

price = df[sale].dropna()
log_price = np.log1p(price)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(
    price, kde=True, bins=40,
    color="#4477AA", ax=axes[0]
)

axes[0].set_title(
    f"Original SalePrice (Skew = {price.skew():.2f})"
)
axes[0].set_xlabel("Sale Price ($)")
axes[0].set_ylabel("Frequency")

sns.histplot(
    log_price, kde=True, bins=40,
    color="#228833", ax=axes[1]
)

axes[1].set_title(
    f"Log-transformed SalePrice (Skew = {log_price.skew():.2f})"
)
axes[1].set_xlabel("Log(SalePrice + 1)")
axes[1].set_ylabel("Frequency")

sns.despine()
plt.tight_layout()

plt.savefig(
    "eda_figures/01_distribution.png",
    dpi=300, bbox_inches="tight"
)
plt.show()
```

Biểu đồ 2 – Correlation Heatmap

```python

numeric = df.select_dtypes(include="number").drop(
    columns=["Id", "PID", "Order", "MS SubClass"],
    errors="ignore"
)

correlations = (
    numeric.corr()[sale]
    .drop(sale)
    .dropna()
    .abs()
    .sort_values(ascending=False)
)

top_features = correlations.head(10).index.tolist()

corr_matrix = numeric[[sale] + top_features].corr()

fig, ax = plt.subplots(figsize=(12, 9))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="vlag",
    center=0,
    linewidths=0.5,
    ax=ax
)

ax.set_title("Correlation with SalePrice")

plt.tight_layout()
plt.savefig(
    "eda_figures/02_correlation.png",
    dpi=300, bbox_inches="tight"
)
plt.show()
```

Biểu đồ 3 – Kitchen Quality vs SalePrice

```python

plot_df = df[[kitchen, sale]].dropna()

order = ["Po", "Fa", "TA", "Gd", "Ex"]
order = [x for x in order if x in plot_df[kitchen].unique()]

fig, ax = plt.subplots(figsize=(9, 6))

sns.boxplot(
    data=plot_df,
    x=kitchen,
    y=sale,
    order=order,
    palette="colorblind",
    ax=ax
)

ax.set_title("SalePrice by Kitchen Quality")
ax.set_xlabel("Kitchen Quality")
ax.set_ylabel("Sale Price ($)")

sns.despine()
plt.tight_layout()

plt.savefig(
    "eda_figures/03_kitchen_quality.png",
    dpi=300, bbox_inches="tight"
)
plt.show()
```

Biểu đồ 4 – Neighborhood Highlight

```python

plot_df = df[[neighborhood, sale]].dropna()

medians = (
    plot_df.groupby(neighborhood)[sale]
    .median()
    .sort_values()
)

top3 = medians.nlargest(3).index
bottom3 = medians.nsmallest(3).index

plot_df["Group"] = np.select(
    [
        plot_df[neighborhood].isin(top3),
        plot_df[neighborhood].isin(bottom3)
    ],
    ["Top 3", "Bottom 3"],
    default="Others"
)

order = medians.index.tolist()

palette = {
    "Top 3": "#228833",
    "Bottom 3": "#CC6677",
    "Others": "#BBBBBB"
}

fig, ax = plt.subplots(figsize=(16, 7))

sns.boxplot(
    data=plot_df,
    x=neighborhood,
    y=sale,
    hue="Group",
    order=order,
    hue_order=["Top 3", "Bottom 3", "Others"],
    palette=palette,
    dodge=False,
    ax=ax
)

ax.set_title("SalePrice Across Neighborhoods")
ax.set_xlabel("Neighborhood")
ax.set_ylabel("Sale Price ($)")
ax.tick_params(axis="x", rotation=60)

ax.legend(title="Neighborhood Group")

sns.despine()
plt.tight_layout()

plt.savefig(
    "eda_figures/04_neighborhood.png",
    dpi=300, bbox_inches="tight"
)
plt.show()
```

Biểu đồ 5 – Multivariate Analysis
