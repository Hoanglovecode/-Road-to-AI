Phụ lục A: Code xác suất cơ bản

```python
import numpy as np

# -----------------------------------------------
# A.1. Mo phong gieo xuc xac 10,000 lan
# -----------------------------------------------
n_trials = 10000
rolls = np.random.randint(1, 7, size=n_trials)

p_even = np.mean(rolls % 2 == 0)
p_greater_than_4 = np.mean(rolls > 4)

print(f"Xac suat so chan (thuc nghiem): {p_even:.4f} (ly thuyet: 0.5)")
print(f"Xac suat lon hon 4 (thuc nghiem): {p_greater_than_4:.4f} (ly thuyet: 0.3333)")

# -----------------------------------------------
# A.2. Mo phong tung 2 dong xu
# -----------------------------------------------
coin1 = np.random.choice(['S', 'N'], size=n_trials)
coin2 = np.random.choice(['S', 'N'], size=n_trials)

p_both_s = np.mean((coin1 == 'S') & (coin2 == 'S'))
p_at_least_one_n = np.mean((coin1 == 'N') | (coin2 == 'N'))

print(f"P(ca hai deu Sap): {p_both_s:.4f} (ly thuyet: 0.25)")
print(f"P(it nhat mot Ngua): {p_at_least_one_n:.4f} (ly thuyet: 0.75)")

# -----------------------------------------------
# A.3. Cong thuc xac suat toan phan
# -----------------------------------------------
P_B = [0.5, 0.3, 0.2]          # Ty le san xuat cua 3 day chuyen
P_A_given_B = [0.02, 0.03, 0.05] # Ty le phe pham cua tung day chuyen

P_A = sum(p_a_b * p_b for p_a_b, p_b in zip(P_A_given_B, P_B))
print(f"P(phe pham) = {P_A:.4f} = {P_A * 100:.1f}%")
```

Phụ lục B: Code công thức Bayes

```python
def bayes_theorem(p_a, p_b_given_a, p_b_given_not_a):
    """
    Tinh xac suat hau nghiem P(A|B) su dung cong thuc Bayes.

    Args:
        p_a: P(A) - Xac suat tien nghiem (Prior)
        p_b_given_a: P(B|A) - Kha nang (Likelihood)
        p_b_given_not_a: P(B|~A) - Ty le bao dong gia

    Returns:
        P(A|B) - Xac suat hau nghiem (Posterior)
    """
    p_not_a = 1.0 - p_a
    # Mau so: Xac suat toan phan cua bien co B
    p_b = p_b_given_a * p_a + p_b_given_not_a * p_not_a
    p_a_given_b = (p_b_given_a * p_a) / p_b
    return p_a_given_b

# Vi du xet nghiem y te
p_disease = 0.0001
p_positive_given_disease = 0.99
p_positive_given_healthy = 0.05

p_disease_given_positive = bayes_theorem(
    p_disease,
    p_positive_given_disease,
    p_positive_given_healthy
)

print(f"P(Mac benh | Xet nghiem duong tinh) = {p_disease_given_positive:.4f} = {p_disease_given_positive * 100:.2f}%")
```

Phụ lục D: Code Multinomial Naive Bayes

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# 1. Tao du lieu mau
data = {
    "text": [
        "Free money now",
        "Call now to claim your prize",
        "Meet me at the park",
        "Lets catch up later",
        "Win a new car today!",
        "Lunch plans?",
        "Congratulations! You won a lottery",
        "Can you send me the report?",
        "Exclusive offer for you",
        "Are you coming to the meeting?"
    ],
    "label": [
        "spam", "spam", "not spam", "not spam", "spam",
        "not spam", "spam", "not spam", "spam", "not spam"
    ]
}

df = pd.DataFrame(data)
df["label_num"] = df["label"].map({"spam": 1, "not spam": 0})

X = df["text"]
y = df["label_num"]

# 2. Chia du lieu
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 3. Vector hoa van ban bang BoW
vectorizer = CountVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

# 4. Huan luyen voi Laplace Smoothing (alpha = 1.0)
model = MultinomialNB(alpha=1.0)
model.fit(X_train_vec, y_train)

# 5. Du doan tren tap test
y_pred = model.predict(X_test_vec)
print(f"Do chinh xac tap test: {accuracy_score(y_test, y_pred) * 100:.2f}%\n")

# 6. Thu nghiem tin nhan tuy y
custom_message = ["Congratulations, you've won a free vacation"]
custom_vec = vectorizer.transform(custom_message)

prediction = model.predict(custom_vec)
probabilities = model.predict_proba(custom_vec)

print(f"Tin nhan: '{custom_message[0]}'")
print(f"Du doan: {'Spam' if prediction[0] == 1 else 'Not Spam'}")
print(f"Xac suat Spam: {probabilities[0][1]:.4f}")
print(f"Xac suat Not Spam: {probabilities[0][0]:.4f}")
```

Phụ lục E: Code chuỗi Markov

```python
import numpy as np

# 1. Ma tran chuyen trang thai P (Nang, Mua, U am)
P = np.array([
    [0.7, 0.1, 0.2],  # Tu Nang
    [0.3, 0.4, 0.3],  # Tu Mua
    [0.4, 0.2, 0.4]   # Tu U am
])

# Kiem tra tinh hop le cua ma tran
print("Kiem tra tong moi hang (phai bang 1):", P.sum(axis=1))

# Trang thai ban dau: Troi Nang -> [1, 0, 0]
initial_state = np.array([1.0, 0.0, 0.0])

# Sau 1 ngay
after_1_day = initial_state @ P
print(f"\nSau 1 ngay: Nang={after_1_day[0]:.3f}, Mua={after_1_day[1]:.3f}, U am={after_1_day[2]:.3f}")

# Sau 2 ngay
after_2_days = initial_state @ np.linalg.matrix_power(P, 2)
print(f"Sau 2 ngay: Nang={after_2_days[0]:.3f}, Mua={after_2_days[1]:.3f}, U am={after_2_days[2]:.3f}")

# Sau 10 ngay
after_10_days = initial_state @ np.linalg.matrix_power(P, 10)
print(f"Sau 10 ngay: Nang={after_10_days[0]:.3f}, Mua={after_10_days[1]:.3f}, U am={after_10_days[2]:.3f}")

# 2. Giai he phuong trinh tim phan phoi dung pi = pi @ P va sum(pi) = 1
A = np.vstack([P.T - np.eye(3), np.ones(3)])
b = np.array([0, 0, 0, 1])
pi_stationary = np.linalg.lstsq(A, b, rcond=None)[0]
print(f"\nPhan phoi dung (Stationary Distribution): {pi_stationary.round(3)}")

# 3. Ham mo phong duong di ngau nhien (Random Walk simulation)
def simulate_markov_chain(P, initial_state, n_steps=20):
    states = ['Nang', 'Mua', 'Uam']
    current = np.random.choice(3, p=initial_state)
    chain = [states[current]]

    for _ in range(n_steps - 1):
        current = np.random.choice(3, p=P[current])
        chain.append(states[current])
    return chain

np.random.seed(42)
simulated_weather = simulate_markov_chain(P, initial_state, 20)
print("\nChuoi thoi tiet mo phong (20 ngay):")
print(" -> ".join(simulated_weather))
```
