```python
import numpy as np

def descriptive_stats(x: np.ndarray) -> dict:
    n=x.shape[0]
    mean_x=np.sum(x)/n
    x=np.sort(x)
    def getmedian(arr:np.ndarray):
        m=np.shape[0]
        if m%2==0:
            return (arr[m//2-1]+arr[m//2])/2
        else:
            return arr[m//2]
    q2=getmedian(x)
    if n%2==0:
        lowerhalf=x[:n//2]
        upperhalf=x[n//2:]
    else:
        lowerhalf=x[:n//2]
        upperhalf=x[n//2+1:]
    q1= getmedian(lowerhalf)
    q3=getmedian(upperhalf)
    iqr=q3-q1
    total=np.sum((x-mean_x)**2)
    variance=total/(n-1)
    std=np.sqrt(variance)
    return {'mean':mean_x,'median':q2,'variance':variance,'std':std,'iqr':iqr}
```

```python
def covariance_matrix(X: np.ndarray) -> np.ndarray:
    N=X.shape[0]
    X_mean=np.sum(X,axis=0)/N
    X_standard=X-X_mean
    return X_standard.T@X_standard/(N-1)
```

```python
def correlation_matrix(X: np.ndarray) -> np.ndarray:
    N,d=X.shape
    cov=covariance_matrix(X)
    B_inverse=np.zeros((d,d))
    for i in range(d):
        B_inverse[i,i]=1/np.sqrt(cov[i,i])
    return B_inverse@cov@B_inverse
```
